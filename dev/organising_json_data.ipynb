```python
import os
import json

# Define the folder containing your JSON files
json_folder = "../docs/figures/data/"  # 🔹 Change this to your actual path

# File extensions for classification
IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg"}
VIDEO_EXTENSIONS = {".mp4"}

def update_customdata(file_path):
    """ Update `customdata` in JSON files to properly format images/videos. """
    try:
        with open(file_path, "r", encoding="utf-8") as file:
            data = json.load(file)

        # Ensure 'data' key exists and is a list
        if "data" not in data or not isinstance(data["data"], list):
            print(f"⚠️ Skipping {file_path} (Invalid format)")
            return

        updated = False  # Track if we modified anything

        for trace in data["data"]:
            if "customdata" in trace and isinstance(trace["customdata"], list):
                updated_customdata = []
                for entry in trace["customdata"]:
                    if isinstance(entry, str):  # Ensure it's a string (URL)
                        if any(entry.lower().endswith(ext) for ext in IMAGE_EXTENSIONS):
                            updated_customdata.append({"image": entry})
                        elif any(entry.lower().endswith(ext) for ext in VIDEO_EXTENSIONS):
                            updated_customdata.append({"video": entry})
                        else:
                            updated_customdata.append({"other": entry})  # If unknown format
                        updated = True  # Mark as modified
                    else:
                        updated_customdata.append(entry)  # Keep original if already formatted

                trace["customdata"] = updated_customdata  # Update `customdata` in place

        # Save the updated JSON file only if changes were made
        if updated:
            with open(file_path, "w", encoding="utf-8") as file:
                json.dump(data, file, indent=4)
            print(f"✅ Updated: {file_path}")
        else:
            print(f"🔹 No changes needed: {file_path}")

    except Exception as e:
        print(f"❌ Error processing {file_path}: {e}")

# Process all JSON files in the folder
for filename in os.listdir(json_folder):
    if filename.endswith("_plot_data.json"):  # Adjust pattern if needed
        file_path = os.path.join(json_folder, filename)
        update_customdata(file_path)

print("🎉 All files processed!")
```
