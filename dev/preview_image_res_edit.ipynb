```python
import cv2
import numpy as np

# Read original image
img = cv2.imread('../docs/images/Figure_1C_full_res.png')

# Calculate new dimensions maintaining aspect ratio
target_width = 600
aspect_ratio = img.shape[1] / img.shape[0]
target_height = int(target_width / aspect_ratio)

# Resize using area interpolation (generally good for downscaling)
low_res = cv2.resize(img, (target_width, target_height), 
                     interpolation=cv2.INTER_AREA)

# Save with moderate compression
cv2.imwrite('../docs/images/Figure_1C_low_res.png', low_res, 
            [cv2.IMWRITE_PNG_COMPRESSION, 6])
```
